# Sentiment Analysis


## 1. Imports

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix
import pandas as pd
```

## 2. Reading the file

```python
dataset = pd.read_csv("data/Tweets_Mg.csv")
dataset.count()
```

## 3. Splitting inputs and outputs

```python
tweets = dataset["Text"].values
classes = dataset["Classificacao"].values
```

## 4. Feature extraction

```python
vectorizer = CountVectorizer(analyzer="word")
freq_tweets = vectorizer.fit_transform(tweets)
```

## 5. Creating training and test sets

```python
# 40/60
x60_train, x60_test, y60_train, y60_test = train_test_split(
    freq_tweets, classes, test_size=0.60, random_state=42
)

# 60/40
x40_train, x40_test, y40_train, y40_test = train_test_split(
    freq_tweets, classes, test_size=0.40, random_state=42
)

# 70/30
x30_train, x30_test, y30_train, y30_test = train_test_split(
    freq_tweets, classes, test_size=0.30, random_state=42
)
```

## 6. Training the models

```python
modelBayes = MultinomialNB()
modelBayes60 = modelBayes.fit(x60_train, y60_train)
modelBayes40 = modelBayes.fit(x40_train, y40_train)
modelBayes30 = modelBayes.fit(x30_train, y30_train)

modelTree = DecisionTreeClassifier()
modelTree60 = modelTree.fit(x60_train, y60_train)
modelTree40 = modelTree.fit(x40_train, y40_train)
modelTree30 = modelTree.fit(x30_train, y30_train)

modelSvc = SVC()
modelSvc60 = modelSvc.fit(x60_train, y60_train)
modelSvc40 = modelSvc.fit(x40_train, y40_train)
modelSvc30 = modelSvc.fit(x30_train, y30_train)
```

## 7. Making predictions

```python
modelBayes60y_predict = modelBayes60.predict(x60_test)
modelBayes40y_predict = modelBayes40.predict(x40_test)
modelBayes30y_predict = modelBayes30.predict(x30_test)

modelTree60y_predict = modelTree60.predict(x60_test)
modelTree40y_predict = modelTree40.predict(x40_test)
modelTree30y_predict = modelTree30.predict(x30_test)

modelSvc60y_predict = modelSvc60.predict(x60_test)
modelSvc40y_predict = modelSvc40.predict(x40_test)
modelSvc30y_predict = modelSvc30.predict(x30_test)
```

## 8. Evaluating the results

```python
from sklearn.metrics import accuracy_score, confusion_matrix

results = {
    "Naive Bayes": [
        ("60%", y60_test, modelBayes60y_predict),
        ("40%", y40_test, modelBayes40y_predict),
        ("30%", y30_test, modelBayes30y_predict),
    ],
    "Decision Tree": [
        ("60%", y60_test, modelTree60y_predict),
        ("40%", y40_test, modelTree40y_predict),
        ("30%", y30_test, modelTree30y_predict),
    ],
    "SVC": [
        ("60%", y60_test, modelSvc60y_predict),
        ("40%", y40_test, modelSvc40y_predict),
        ("30%", y30_test, modelSvc30y_predict),
    ],
}

for model_name, tests in results.items():
    print(model_name.upper())
    print("\n")
    
    for split_name, y_true, y_pred in tests:
        print(f"TEST WITH {split_name} OF THE DATA")
        accuracy = accuracy_score(y_true, y_pred)
        print("accuracy:", accuracy)
        print("confusion matrix:")
        print(confusion_matrix(y_true, y_pred))
        print()
    
    print("#############################################################\n")
```

```python
print("\nSample predictions:")

sample_texts = [
    "esse produto é muito bom",
    "não gostei da experiência",
    "mais ou menos"
]

X_sample = vectorizer.transform(sample_texts)
predictions = modelSvc30.predict(X_sample)

for text, pred in zip(sample_texts, predictions):
    print(text, "->", pred)
```
